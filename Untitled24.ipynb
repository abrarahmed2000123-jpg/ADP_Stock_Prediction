```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
df = pd.read_csv("/content/Mall_Customers.csv")

```

```python
#Viev first 5 rows of the Dataset
df.head()
```

```python
#Check column types and null
df.info()
```

```python
#Numerical columns
df.describe()
```

```python
#Checking missing values
df.isnull().sum()
```

```python
# Droping CustomerID
df = df.drop("CustomerID", axis=1)
```

```python
# Converting the columns in numerics
df['Gender'] = df['Gender'].map({'Male': 0, 'Female': 1})
```

```python
# Selecting features for clustering
# We would focus on Annual Income and Spending Score
X = df[['Annual Income (k$)', 'Spending Score (1-100)']]
```

```python
# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python

wcss = []

# Try 1 to 10 clusters
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
```

```python

# Plot the Elbow Graph
plt.figure(figsize=(8,5))
plt.plot(range(1,11), wcss, marker='o')
plt.title('Elbow Method - Find Optimal Clusters')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.show()
```

“The Elbow Method shows 4 optimal clusters, grouping customers by income and spending for better targeting.”

```python
# Assume elbow shows 5 clusters
kmeans = KMeans(n_clusters=5, random_state=42)
df['Cluster'] = kmeans.fit_predict(X_scaled)

# Check cluster assignment for first 10 customers
print(df.head(10))
```

```python
# Silhouette Score: closer to 1 = better clustering
score = silhouette_score(X_scaled, df['Cluster'])
print("Silhouette Score:", score)

# Check cluster sizes
print(df['Cluster'].value_counts())
```

```python
# Visualizations
plt.figure(figsize=(8,6))
sns.scatterplot(
    x=df['Annual Income (k$)'],
    y=df['Spending Score (1-100)'],
    hue=df['Cluster'],
    palette='Set2',
    s=100,   # size of points
    alpha=0.8
)
plt.title('Customer Segments (KMeans)')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend(title='Cluster')
plt.show()
```

“The scatter plot shows 4 distinct customer groups based on income and spending patterns.”

```python
# Calculate mean values of each cluster
cluster_summary = df.groupby('Cluster').mean()
print("Cluster Summary (Mean Values):")
print(cluster_summary)

# Optional: count of customers per cluster
cluster_counts = df['Cluster'].value_counts()
print("\nNumber of Customers per Cluster:")
print(cluster_counts)

# Cluster Insights
print("\nCluster Insights:")
print("Cluster 0: Medium income and spending — average customers (largest group).")
print("Cluster 1: High income and high spending — premium buyers.")
print("Cluster 2: Low income and high spending — young impulsive buyers.")
print("Cluster 3: High income and low spending — careful spenders.")
print("Cluster 4: Low income and low spending — budget customers.")
```

“Customer segmentation using KMeans identified five distinct customer groups based on income and spending behavior. These insights can help the business design targeted marketing strategies and improve customer engagement.”
